**Level:** Beginner  

## Objective
To analyse sales data, identify product performance,
calculate business metrics, and generate useful insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load sales dataset
df = pd.read_csv("../sales_data.csv")

# Display first 5 rows
df.head()
```

```python
# Dataset dimensions
print("Rows and Columns:", df.shape)

# Dataset information
df.info()

# Statistical summary
df.describe()
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# Check unique products
print("\nUnique Products:", df["Product"].nunique())

# Check unique regions
print("\nUnique Regions:", df["Region"].nunique())
```

```python
# Calculate profit
df["Profit"] = df["Sales"] - df["Cost"]

# Display sales and profit
df[["Product", "Sales", "Cost", "Profit"]]
```

```python
# Total sales by product
product_sales = df.groupby("Product")["Sales"].sum()

# Sort products by sales
product_sales = product_sales.sort_values(ascending=False)

print("Product-wise Sales:")
display(product_sales)
```

```python
# Calculate region-wise sales
region_sales = df.groupby("Region")["Sales"].sum()

# Sort sales in descending order
region_sales = region_sales.sort_values(ascending=False)

print("Region-wise Sales:")
display(region_sales)
```

```python
# Product-wise sales visualization

plt.figure(figsize=(10, 5))

product_sales.plot(kind="bar", title="Product-wise Sales Analysis")

plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(
    kind="bar",
    title="Region-wise Sales Analysis"
)

plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
total_sales = df["Sales"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()
total_orders = len(df)

print("BUSINESS PERFORMANCE SUMMARY")
print("----------------------------")
print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Total Cost: ₹{total_cost:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Profit Margin: {(total_profit / total_sales) * 100:.2f}%")
```

## Business Insights

1. Product-wise sales analysis identifies the best-performing products.
2. Region-wise sales analysis helps compare regional performance.
3. Profit calculation helps understand business profitability.
4. Business KPIs provide an overview of total sales, cost, profit and orders.
5. Visualizations make sales trends and comparisons easier to understand.
